```python
import torch
from easysteer.steer import (
    StatisticalControlVector,
    extract_statistical_control_vector,
    extract_diffmean_control_vector,
    extract_pca_control_vector,
    extract_lat_control_vector,
    extract_linear_probe_control_vector
)
```

```python
all_hidden_states = torch.load("qwen2_5VL_all_hidden_states_tensor.pt")
```

```python
positive_indices = [i for i in range(all_hidden_states.shape[0]) if i % 2 == 1]
negative_indices = [i for i in range(all_hidden_states.shape[0]) if i % 2 == 0]
```

```python
control_vector = extract_diffmean_control_vector(
    all_hidden_states=all_hidden_states.unsqueeze(dim=2),  
    positive_indices=positive_indices,     # Indices of positive samples
    negative_indices=negative_indices,     # Indices of negative samples
    model_type="qwen2_5vl",  
    token_pos=-1,      # Use the last token 
    normalize=False,
)
control_vector.export_gguf("qwen2_5_100_pair_diffmean_wo_norm.gguf")
```
